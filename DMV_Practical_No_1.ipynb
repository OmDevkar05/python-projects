```python
# Fill object columns (text) with mode and numeric columns with mean
for col in df_json.columns:
    if df_json[col].isnull().any():
        if df_json[col].dtype == 'object':
            df_json[col].fillna(df_json[col].mode()[0], inplace=True)
        else:
            df_json[col].fillna(df_json[col].mean(), inplace=True)

# Check if any missing values remain
df_json.isnull().sum()
```

```python
df= df.drop_duplicates()
df_excel = df_excel.drop_duplicates()
df_json = df_json.drop_duplicates()
```

```python
df['Source'] = 'CSV'
df_excel['Source'] = 'Excel'
df_json['Source'] = 'JSON'
```

```python
unified_df = pd.concat([df, df_excel, df_json], ignore_index=True)
```

```python
print("Unified dataframe shape:", unified_df.shape)
unified_df.head()
```

```python
unified_df['Order Date'] = pd.to_datetime(unified_df['Order Date'], errors='coerce')
unified_df['Year'] = unified_df['Order Date'].dt.year
unified_df['Month'] = unified_df['Order Date'].dt.month
unified_df['Day'] = unified_df['Order Date'].dt.day
```

```python
unified_df['Amount'] = unified_df['Quantity'] * unified_df['Unit price']
unified_df['Profit Margin %'] = (unified_df['Profit'] / unified_df['Sales']) * 100
```

```python
unified_df.describe()
```

```python
total_sales = unified_df['Sales'].sum()
print("Total Sales:", total_sales)
```

```python
aov = unified_df['Sales'].mean()
print("Average Order Value:", aov)
```

```python
total_profit = unified_df['Profit'].sum()
print("Total Profit:", total_profit)
```

```python
if 'Category' in unified_df.columns:
    print(unified_df.groupby('Category')['Sales'].sum())
```

```python
if 'Region' in unified_df.columns:
    print(unified_df.groupby('Region')['Sales'].sum())
```

```python
if 'Month' in unified_df.columns:
    print(unified_df.groupby('Month')['Sales'].sum())
```

```python
import matplotlib.pyplot as plt
unified_df.groupby('Category')['Sales'].sum().plot(kind='bar', color='teal')
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
unified_df['Category'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title("Product Category Distribution")
plt.ylabel("")
plt.show()
```

```python
unified_df.groupby('Region')['Sales'].sum().plot(kind='bar', color='orange')
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.show()
```

```python
if 'Month' in unified_df.columns:
    unified_df.groupby('Month')['Sales'].sum().plot(kind='line', marker='o', color='purple')
    plt.title("Monthly Sales Trend")
    plt.xlabel("Month")
    plt.ylabel("Sales")
    plt.grid(True)
    plt.show()
```
